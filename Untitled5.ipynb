```python
from scipy.stats import chi2_contingency

def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum()
    r, k = confusion_matrix.shape
    
    return np.sqrt(
        (chi2 / n) /
        min(k - 1, r - 1)
    )


# Test Contract × Churn
contract_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

chi2, p_value, dof, expected = chi2_contingency(contract_table)

v = cramers_v(contract_table.values)

print("Contingency Table:")
display(contract_table)

print(f"\nChi-square statistic: {chi2:.2f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.6g}")
print(f"Cramér's V: {v:.3f}")
```

```python
# Statistical association between categorical features and churn

categorical_features = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]

results = []

for feature in categorical_features:
    
    contingency_table = pd.crosstab(
        df[feature],
        df["Churn"]
    )
    
    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )
    
    v = cramers_v(contingency_table.values)
    
    results.append({
        "feature": feature,
        "chi2": chi2,
        "p_value": p_value,
        "cramers_v": v
    })

association_results = (
    pd.DataFrame(results)
      .sort_values("cramers_v", ascending=False)
      .reset_index(drop=True)
)

association_results
```

```python
# Create modeling dataset
model_df = df.copy()

# Remove target leakage / non-predictive identifiers
model_df = model_df.drop(
    columns=["customerID", "tenure_group", "monthly_charge_band"]
)

# Encode target
model_df["Churn"] = model_df["Churn"].map({"No": 0, "Yes": 1})

print("Modeling dataset shape:", model_df.shape)
print("\nTarget distribution:")
print(model_df["Churn"].value_counts())
print("\nTarget rate:")
print(f"Churn rate: {model_df['Churn'].mean() * 100:.2f}%")
```

```python
from sklearn.model_selection import train_test_split

X = model_df.drop(columns=["Churn"])
y = model_df["Churn"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining churn rate:")
print(f"{y_train.mean() * 100:.2f}%")

print("\nTest churn rate:")
print(f"{y_test.mean() * 100:.2f}%")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]

categorical_features = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore", drop="first"),
            categorical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)

logistic_pipeline.fit(X_train, y_train)

print("Logistic Regression baseline trained successfully.")
```

```python
def evaluate_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]

    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_proba)
    }


baseline_results = pd.DataFrame([
    evaluate_model(
        logistic_pipeline,
        X_test,
        y_test,
        "Logistic Regression"
    )
])

baseline_results
```

```python
from sklearn.metrics import precision_recall_curve

y_proba_tuned = best_xgb.predict_proba(X_test)[:, 1]

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_proba_tuned
)

threshold_results = pd.DataFrame({
    "threshold": thresholds,
    "precision": precision[:-1],
    "recall": recall[:-1]
})

threshold_results["f1"] = (
    2 * threshold_results["precision"] * threshold_results["recall"] /
    (
        threshold_results["precision"] +
        threshold_results["recall"]
    )
)

threshold_results.sort_values(
    "f1",
    ascending=False
).head(10)
```

```python
from sklearn.model_selection import cross_val_predict

oof_proba = cross_val_predict(
    best_xgb,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

oof_precision, oof_recall, oof_thresholds = precision_recall_curve(
    y_train,
    oof_proba
)

oof_threshold_results = pd.DataFrame({
    "threshold": oof_thresholds,
    "precision": oof_precision[:-1],
    "recall": oof_recall[:-1]
})

oof_threshold_results["f1"] = (
    2 * oof_threshold_results["precision"] * oof_threshold_results["recall"] /
    (
        oof_threshold_results["precision"] +
        oof_threshold_results["recall"]
    )
)

best_oof_threshold = (
    oof_threshold_results
    .loc[oof_threshold_results["f1"].idxmax(), "threshold"]
)

best_oof_metrics = (
    oof_threshold_results
    .loc[oof_threshold_results["f1"].idxmax()]
)

print(f"Optimal threshold from training CV: {best_oof_threshold:.4f}")
print(f"OOF Precision: {best_oof_metrics['precision']:.4f}")
print(f"OOF Recall:    {best_oof_metrics['recall']:.4f}")
print(f"OOF F1:        {best_oof_metrics['f1']:.4f}")
```

```python
# Evaluate the retention-prioritization list at the locked OOF threshold
retention_threshold = 0.5416

y_test_pred_priority = (test_probabilities >= retention_threshold).astype(int)

high_risk_count = y_test_pred_priority.sum()
actual_churners = y_test.sum()
captured_churners = ((y_test_pred_priority == 1) & (y_test == 1)).sum()

print("Retention Prioritization at OOF-Selected Threshold")
print("-" * 50)
print(f"Threshold: {retention_threshold:.4f}")
print(f"Customers flagged: {high_risk_count:,} / {len(y_test):,}")
print(f"Flagged customer share: {high_risk_count / len(y_test):.2%}")
print(f"Actual churners in test set: {actual_churners:,}")
print(f"Churners captured: {captured_churners:,}")
print(f"Churn capture rate: {captured_churners / actual_churners:.2%}")
print(f"Precision of retention list: {captured_churners / high_risk_count:.2%}")
```

```python
# Build customer-level retention prioritization table
retention_list = X_test.copy()

retention_list["CustomerID"] = df.loc[X_test.index, "customerID"]
retention_list["Churn_Probability"] = test_probabilities
retention_list["Risk_Flag"] = np.where(
    retention_list["Churn_Probability"] >= retention_threshold,
    "High Risk",
    "Lower Risk"
)
retention_list["Actual_Churn"] = y_test.values

# Add useful business attributes
retention_list = retention_list[
    [
        "CustomerID",
        "Churn_Probability",
        "Risk_Flag",
        "tenure",
        "Contract",
        "InternetService",
        "OnlineSecurity",
        "TechSupport",
        "PaymentMethod",
        "MonthlyCharges",
        "Actual_Churn"
    ]
].sort_values(
    "Churn_Probability",
    ascending=False
).reset_index(drop=True)

print("Retention prioritization list created.")
print(f"Total customers: {len(retention_list):,}")
print(f"High-risk customers: {(retention_list['Risk_Flag'] == 'High Risk').sum():,}")

display(retention_list.head(10).style.format({
    "Churn_Probability": "{:.2%}",
    "MonthlyCharges": "{:.2f}"
}))
```

## Business Recommendations

### 1. Prioritize early-tenure customers
Customers in the 1–6 month tenure cohort show a **53.33% churn rate**, approximately twice the overall **26.54% baseline**. Retention efforts should therefore focus heavily on early lifecycle customers through onboarding, proactive support, and early engagement interventions.

### 2. Prioritize month-to-month customers
Month-to-month customers have a **42.71% churn rate**, compared with **11.27%** for one-year and **2.83%** for two-year contracts. Where commercially appropriate, retention teams should prioritize contract-renewal and longer-term-plan conversion opportunities.

### 3. Focus on high-risk customer segments
The three largest high-risk segments identified earlier represent **34.29% of customers but 69.07% of observed churners**. These segments provide a concentrated starting point for targeted retention rather than broad customer-wide campaigns.

### 4. Investigate fiber + month-to-month customers
The **fiber optic + month-to-month** combination has a **54.61% churn rate**. The highest-risk multidimensional segment—month-to-month + fiber optic + no Online Security + no Tech Support—has a **60.70% churn rate**. These customers warrant targeted investigation and retention treatment.

### 5. Use model-driven retention prioritization
At the OOF-selected probability threshold of **0.5416**, the model flags **38.89% of customers** while capturing **76.20% of observed churners**, producing **1.96× targeting lift** versus targeting the same customer share randomly.

### 6. Differentiate intervention intensity by risk tier
The risk-tier framework enables differentiated treatment:
- **High Risk:** intensive retention intervention
- **Medium Risk:** proactive engagement and targeted offers
- **Low Risk:** standard engagement; avoid unnecessary retention spend

### Analytical caveat
These findings represent observational associations and model-based predictions. They should be validated through controlled retention experiments before treating any intervention as causal.

Three classification approaches—Logistic Regression, Random Forest, and XGBoost—were evaluated using stratified train/test splitting. XGBoost was further optimized using cross-validated hyperparameter search, with **Average Precision** as the tuning objective due to class imbalance.

The final tuned XGBoost model achieved **0.8442 ROC-AUC** on the held-out test set.

### Retention Prioritization
Rather than optimizing only for default classification accuracy, the decision threshold was selected using **out-of-fold predictions** to prioritize churn detection.

At the locked threshold of **0.5416**:

- **76.20% recall** — captured 285 of 374 observed churners
- **52.01% precision** among flagged customers
- **38.89% of customers targeted**
- **1.96× retention targeting lift**

### Key Churn Drivers

The highest-risk multidimensional segment—**month-to-month + fiber optic + no Online Security + no Tech Support**—showed a **60.70% observed churn rate**.

### Limitations

1. **Observational data:** Relationships identified in the dataset represent associations and should not be interpreted as causal effects.
2. **Historical dataset:** Model performance may change as customer behavior, pricing, products, and market conditions evolve.
3. **No intervention outcomes:** The analysis estimates churn risk but does not measure whether a retention action actually prevents churn.
4. **Business costs not provided:** A production retention strategy should incorporate intervention cost, customer value, and expected retention benefit when setting the final decision threshold.
5. **External validation:** Before deployment, the model should be evaluated on a temporally separated or independently collected dataset.
